```python
import pandas as pd
import matplotlib.pyplot as plt
# -----------------------------------------
Archivo: funciones_analisis.py
# -----------------------------------------
# Función: cargar_datos
# Librería: pandas
# Descripción: carga un archivo Excel y retorna un DataFrame
# Reutilizable para cualquier proyecto con datos tabulares
# -----------------------------------------
def cargar_datos(ruta_archivo):
return pd.read_excel(ruta_archivo)

# -----------------------------------------
# Función: calcular_metricas_por_medio
# Librería: pandas

# Descripción: agrupa los datos por medio publicitario y
# calcula métricas promedio relevantes para análisis comparativo
# -----------------------------------------
def calcular_metricas_por_medio(df):
return df.groupby("medio")[[
"impresiones",
"clics",
"conversiones",
"roi"
]].mean()

# -----------------------------------------
# Función: exportar_reportes_excel
# Librería: pandas
# Descripción: genera reportes diferenciados para áreas de la empresa
# -----------------------------------------
def exportar_reportes_excel(df, resumen):
resumen[["roi", "conversiones"]].to_excel(
"reporte_area_estrategica.xlsx"
)
df[[
"campaña_id",
"medio",
"presupuesto_usd",
"clics"
]].to_excel(
"reporte_area_operativa.xlsx",
index=False
)
```
