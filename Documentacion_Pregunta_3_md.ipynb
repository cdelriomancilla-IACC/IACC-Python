# Analisis de Calidad de Servicio - RedConecta Telecom S.A.

## 1. Objetivos del Proyecto
* Analizar la información operacional de la empresa para respaldar decisiones técnicas orientadas a mejorar la calidad del servicio.
* Identificar cuellos de botella en la atención de soporte técnico mediante el cruce de tickets y tiempos de resolución.
* Evaluar el impacto de la degradación de red (latencia y fallas) en la experiencia de los clientes y el cumplimiento de SLA.
* Generar salidas de datos específicas y un panel visual reproducible para monitoreo periódico.

## 2. Estructura del Análisis
1. **Extracción y Carga de Datos:** Lectura del archivo oficial `dataset_set_D_telecomunicaciones.xlsx` mediante Pandas, asegurando la integridad del archivo original.
2. **Separación y Salidas Operativas (ETL):** Generación de dos archivos CSV independientes:
   * `salida_soporte_tecnico.csv`: Enfocado en volumen de fallas, tickets y horas de resolución.
   * `salida_experiencia_clientes.csv`: Enfocado en clientes afectados, latencia y cumplimiento de SLA.
3. **Exploración Visual:** Construcción de un dashboard de 4 paneles en Matplotlib (latencia, dispersión de tickets vs. resolución, clientes por servicio y serie temporal de incidencias).
4. **Control de Versiones:** Registro y trazabilidad de cambios mediante GitHub.

## 3. Principales Resultados
* **Equivalencia de latencia por zona:** La latencia promedio se mantiene sobre los 60 ms en todas las regiones, registrándose el mayor valor en la Región Centro (~68 ms).
* **Tiempos de resolución dispersos:** No se observa una relación directa entre cantidad de tickets y tiempo de resolución; existen fallas que tardan más de 40 horas en resolverse incluso con pocos tickets.
* **Afectación transversal:** El volumen promedio de clientes afectados se mantiene constante entre los servicios de Empresas, Internet Hogar y Móvil (entre 260 y 270 clientes por evento).
* **Cumplimiento de SLA:** Más del 50% de los registros operacionales incumplen los compromisos de servicio, requiriendo planes de contingencia técnica inmediatos.
