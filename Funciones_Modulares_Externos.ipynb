```python
%%writefile funciones_red.py
import joblib
import numpy as np
from scipy import stats


def calcular_metricas(serie_datos):
  # Conversión a arreglo NumPy y cálculos directos
  arreglo = np.array(serie_datos)
  media = np.mean(arreglo)
  desviacion = np.std(arreglo)

  # Cálculo estadístico con SciPy
  asimetria = stats.skew(arreglo)

  return {
      'promedio': round(media, 2),
      'desviacion': round(desviacion, 2),
      'asimetria': round(asimetria, 2),
  }


def guardar_resultados(datos, nombre_archivo):
  # Guardado directo en disco con Joblib
  joblib.dump(datos, nombre_archivo)
```

```python
%%writefile analisis.py
from funciones_red import calcular_metricas, guardar_resultados
import pandas as pd

# Lectura directa del dataset
df = pd.read_excel('dataset_set_D_telecomunicaciones.xlsx')

# Llamada a la función externa
metricas = calcular_metricas(df['latencia_ms'])

# Mostrar en pantalla de forma simple y directa (sin for)
print('Resultados de latencia:', metricas)

# Guardar con Joblib
guardar_resultados(metricas, 'metricas_latencia.joblib')
```
